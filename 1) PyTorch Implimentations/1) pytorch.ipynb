# Creating a Tensor

```python
# using empty 
tensor = torch.empty(2,3) # creates a 2x3 matrix with random values
# it creates an empty tensor and the values which it returns are the previously stored values in the memory location. So it is not initialized with any values.
print(tensor)
```

```python
# using zeros
tensor = torch.zeros(2,3) # creates a 2x3 matrix with all values as 0
print(tensor)
```

```python
# using ones
tensor = torch.ones(2,3) # creates a 2x3 matrix with all values as 1
print(tensor)
```

```python
# using eye
tensor = torch.eye(3) # creates a 3x3 identity matrix
print(tensor)
```

```python
# using rand 
tensor = torch.rand(2,3) # creates a 2x3 matrix with random values between 0 and 1
print(tensor) 
"""
     the problem with rand is that it generats random values every time we run it.
     so if we want to generate the same random values every time we run it, we can use the seed function.

"""
```

```python
# using seed 
torch.manual_seed(100) # sets the seed for generating random numbers
tensor = torch.rand(2,3) # creates a 2x3 matrix with random values between 0 and 1
print(tensor) 
```

```python
torch.manual_seed(100) # sets the seed for generating random numbers
tensor = torch.rand(2,3) # creates a 2x3 matrix with same random values 
print(tensor)
```

```python
# using tensor 
tensor = torch.tensor([[1,2,3],[4,5,6]]) # creates a 2x3 matrix with the given values
print(tensor)
```

### some other meathodes


```python
# using arange
tensor = torch.arange(0,10,2) # creates a 1D tensor with values from 0 to 10 with a step of 2
print(tensor)
# using linspace
tensor = torch.linspace(0,10,5) # creates a 1D tensor with 5 values from 0 to 10
print(tensor)
# using full
tensor = torch.full((2,3),7) # creates a 2x3 matrix with all values as 7
print(tensor)
```

# Reduction 

```python
torch.manual_seed(30)
e = torch.randint(size = (2,3), low = 0, high = 10, dtype= torch.float)
print(e)
# sum 
print("Sum", torch.sum(e), e.sum())

# sum along columns
print("Sum alog conumns keeping the dimension",torch.sum(e, dim = 0, keepdim= True))

# sum along rows 
print("Sum along Rows: ", torch.sum(e, dim = 1, keepdim= True), e.sum(dim=1, keepdim= True), sep="\n")
```

```python
# mean median
print("Sum", torch.mean(e), e.mean())

# mean along columns
print("mean alog conumns keeping the dimension",torch.mean(e, dim = 0, keepdim= True))

# mean along rows 
print("mean along Rows: ", torch.mean(e, dim = 1, keepdim= True), e.mean(dim=1, keepdim= True), sep="\n")
```

```python
# max min 
print(e.max())
print()
print(e.max(dim=0, keepdim= True))
print()
print(e.max(dim=1, keepdim=True))
```

```python
# std
print(torch.std(e))

# variance
torch.var(e)
```

```python
torch.argmax(e)
```

```python
torch.argmin(e)
```

# Pushing tensor on gpu

```python
x = torch.randn(3, 4)

print("Before moving:", x.device)

# 1. Define the device (uses GPU if available, otherwise CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 2. Move your data (your existing code)
x = x.to(device) # or x = x.cuda() if torch.cuda.is_available() else x.cpu()

print("After moving:", x.device)
```

```python
import time
# Define the size of the matrices
size = 10000 # Large size for performance comparison
# Create random matrices on CPU
matrix_cpu1 = torch.randn(size,
size)
matrix_cpu2 = torch.randn(size,
size)
# Measure time on CPU
start_time = time.time()
result_cu = torch.matmul(matrix_cpu1, matrix_cpu2) # Matrix multiplication on CPU
cpu_time = time. time () - start_time
print (f"Time on CPU: {cpu_time:.4f} seconds")
# Move matrices to GPU
matrix_gpu1 = matrix_cpu1.to('cuda')
matrix_gpu2 = matrix_cpu2.to('cuda')
# Measure time on GPU
start_time = time.time()
result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2) # Matrix multiplication on GPU
torch.cuda.synchronize()
# Ensure all GPU operations are complete
gpu_time = time.time() - start_time
```

```python
print(gpu_time)
```
