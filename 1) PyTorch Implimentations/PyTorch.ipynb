# Creating a Tensor

```python
# using empty 
tensor = torch.empty(2,3) # creates a 2x3 matrix with random values
# it creates an empty tensor and the values which it returns are the previously stored values in the memory location. So it is not initialized with any values.
print(tensor)
```

```python
# using zeros
tensor = torch.zeros(2,3) # creates a 2x3 matrix with all values as 0
print(tensor)
```

```python
# using ones
tensor = torch.ones(2,3) # creates a 2x3 matrix with all values as 1
print(tensor)
```

```python
# using eye
tensor = torch.eye(3) # creates a 3x3 identity matrix
print(tensor)
```

```python
# using rand 
tensor = torch.rand(2,3) # creates a 2x3 matrix with random values between 0 and 1
print(tensor) 
"""
     the problem with rand is that it generats random values every time we run it.
     so if we want to generate the same random values every time we run it, we can use the seed function.

"""
```

```python
# using seed 
torch.manual_seed(100) # sets the seed for generating random numbers
tensor = torch.rand(2,3) # creates a 2x3 matrix with random values between 0 and 1
print(tensor) 
```

```python
torch.manual_seed(100) # sets the seed for generating random numbers
tensor = torch.rand(2,3) # creates a 2x3 matrix with same random values 
print(tensor)
```

```python
# using tensor 
tensor = torch.tensor([[1,2,3],[4,5,6]]) # creates a 2x3 matrix with the given values
print(tensor)
```

### some other meathodes


```python
# using arange
tensor = torch.arange(0,10,2) # creates a 1D tensor with values from 0 to 10 with a step of 2
print(tensor)
# using linspace
tensor = torch.linspace(0,10,5) # creates a 1D tensor with 5 values from 0 to 10
print(tensor)
# using full
tensor = torch.full((2,3),7) # creates a 2x3 matrix with all values as 7
print(tensor)
```

#### Finding the shape of tensor

```python
x = torch.tensor([1,2,3])
print(x.shape)
```

```python
# creating a new tensor with the same shape as x
y = torch.zeros_like(x) # creates a tensor with the same shape as x and all values as 0
print(y)
y = torch.ones_like(x) # creates a tensor with the same shape as x and all values as 1
print(y)
#y = torch.rand_like(x) # creates a tensor with the same shape as x and random values between 0 and 1
#print(y)
y = torch.full_like(x,7) # creates a tensor with the same shape as x and all values as 7
print(y)
#y = torch.eye_like(x) # creates a tensor with the same shape as x and all values as 1 in the diagonal and 0 elsewhere
#print(y)
#y = torch.arange_like(x) # creates a tensor with the same shape as x and values from 0 to the number of elements in x
#print(y)
y = torch.empty_like(x) # creates a tensor with the same shape as x and all values as 1 in the diagonal and 0 elsewhere
print(y)

"""
    # We will see why rand_like, arrange_like and eye_like are not working
    # it is because they generate float values where as x is an integer data type tensor
    # So we explecetly need to tell that we want the float type values
"""
```

```python
x = torch.tensor([1, 2, 3]) # Starts as int64

# Reassign the result to update x
x = x.to(torch.float32)

print("x:   ", x, "\n", x.dtype)
# Output:
# x:    tensor([1., 2., 3.]) 
# torch.float32
```

```python
# soln to our previous problem
y = torch.rand_like(x, dtype = torch.float32) # we just need to specify the data type
print(y)
```

# Operations on tensor

### 1) scalar operation

```python
# addition
x = torch.ones(2,3)
x
```

```python
print(x + 2)
print( x - 1)
print(x * 2)
print(x / 2)
print(x // 2)
print(x % 2)
```

```python
# operation between two tensor
t1 = torch.full((2,2), 8)
t2 = torch.full((2,2),2)
print(t1)
print(t2)
```

```python
# add 
print(t1 + t2)
# sub 
print(t1 - t2)
# mul
print(t1 * t2)
# div
print(t1/t2)
# power 
print(t1**t2)
# mod 
print(t1%t2)
```
