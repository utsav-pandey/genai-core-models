## Problems with batch GD
1) Computationaly costly (Memory inefficient)
    - need to keep the complete data in ram 
2) may not do better convergence
    - because we need to work on complete before updating parameters once so the update is slow
3) Solution - using batches of data to train the model
    - just work sections of dataset one by one (use mini-batch GD) 

## can't we make our own algo from scratch
- we can but it has following problems
    1) No standard interface for data
        - the dataset may not available in form of tensors
        - there is no way to import data
    2) No easy way to apply transformations
        - can't add transformations before every epoch
    3) Shuffling and sampling
        - no way to shuffl and sample from data
    4) Batch management & Parallelization
